**Langkah Praktikum**

```python
import pandas as pd
import numpy as np

# Dataset karyawan (sengaja mengandung masalah kualitas)
data = {
    'usia'      : [25, 32, np.nan, 45, 28, 51, 38, np.nan, 29, 41],
    'pendapatan': [4_000_000, 7_500_000, 5_200_000, 12_000_000, 4_800_000,
                   15_000_000, 8_100_000, 6_300_000, np.nan, 9_900_000],
    'pendidikan': ['SMA','S1','SMA','S2','SMA','S2','S1','S1','SMA','S2'],  # ordinal
    'kota'      : ['Bandung','Jakarta','Bandung','Surabaya','Jakarta',
                   'Surabaya','Jakarta','Bandung','Bandung','Jakarta'],    # nominal
    'membeli'   : ['Tidak','Ya','Tidak','Ya','Tidak','Ya','Ya','Tidak','Tidak','Ya']  # target
}
df = pd.DataFrame(data)
df
```

Langkah 1 — Menangani nilai hilang (imputasi median/modus).

```python
df['usia'] = df['usia'].fillna(df['usia'].median())
df['pendapatan'] = df['pendapatan'].fillna(df['pendapatan'].median())
```

Langkah 2 — Memisahkan fitur (X) dan label (y).

```python
X = df.drop(columns=['membeli'])
y = df['membeli']
```

Langkah 3 — Encoding kategorikal (ordinal & one-hot).

```python
X['pendidikan'] = X['pendidikan'].map({'SMA':0,'S1':1,'S2':2}) # ordinal
X = pd.get_dummies(X, columns=['kota'], dtype=int) # nominal
y = y.map({'Tidak':0,'Ya':1})
```

Langkah 4 — Membagi data (SEBELUM scaling).

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(
X, y, test_size=0.3, random_state=42)
```

Langkah 5 — Penskalaan (fit pada latih, transform pada latih & uji).

```python
from sklearn.preprocessing import StandardScaler
num = ['usia','pendapatan','pendidikan']
sc = StandardScaler()
X_train[num] = sc.fit_transform(X_train[num])
X_test[num] = sc.transform(X_test[num]) # cegah data leakage
```

Langkah 6 — Verifikasi: tidak ada nilai hilang, semua numerik, skala seragam.

```python
# Cek apakah masih ada nilai yang hilang
print("Jumlah nilai hilang:")
print(X_train.isnull().sum())
print()

# Cek tipe data setiap kolom
print("Tipe data:")
print(X_train.dtypes)
print()

# Melihat statistik data setelah scaling
print("Statistik data setelah scaling:")
print(X_train.describe())
```

4. Jelaskan mengapa fit_transform pada latih tetapi hanya transform pada uji.

Karena scaler harus belajar dari data latih saja agar tidak terjadi data leakage. Data uji tidak ikut serta untuk menentukan rata-rata standar deviasi, nilai minimum/maksimum. Jadi data uji tetap menjadi data yang benar-benar "baru" untuk menguji model.

**E. Refleksi**

1. Mengapa urutan 'split dulu, baru scaling' penting?

Karena Urutan tersebut penting agar data uji tidak ikut memengaruhi proses scaling pada data latih. Jika scaling dilakukan sebelum data dibagi, informasi dari data uji dapat ikut digunakan dalam perhitungan nilai seperti mean dan standar deviasi. Hal ini disebut data leakage dan dapat membuat hasil evaluasi model menjadi kurang objektif. Oleh karena itu, data dibagi terlebih dahulu, kemudian scaling dilakukan berdasarkan data latih dan diterapkan pada data uji.

2. Kapan label encoding dan kapan one-hot encoding?

Label encoding digunakan ketika kategori memiliki urutan atau tingkatan tertentu, misalnya rendah, sedang, tinggi atau untuk mengubah label target seperti Lulus dan Tidak Lulus menjadi angka. Sedangkan one-hot encoding digunakan untuk data kategori yang tidak memiliki urutan, misalnya Jawa Barat, Jawa Tengah, Jawa Timur. Setiap kategori dibuat menjadi kolom tersendiri dengan nilai 0 atau 1 sehingga model tidak menganggap ada hubungan tingkatan antar kategori.

3. Apa perbedaan normalisasi dan standardisasi?

Normalisasi mengubah nilai data ke rentang tertentu, biasanya 0 sampai 1, sehingga skala setiap fitur menjadi lebih seragam. Sementara standardisasi mengubah data berdasarkan rata-rata (mean) dan standar deviasi sehingga data memiliki mean 0 dan standar deviasi 1. Normalisasi cocok ketika ingin membatasi nilai pada rentang tertentu, sedangkan standardisasi sering digunakan ketika model membutuhkan data dengan skala yang berpusat di sekitar nol.
